```python
import pandas as pd
```

```python
df = pd.read_csv('../data/spipi_car_sales.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.insert(0, 'Sale_ID', range(1, len(df) + 1))
df.drop(columns=['Car_id'], inplace=True)
```

```python
df.head()
```

```python
print(df.dtypes[['Sale_ID']])
print(df.shape)
```

```python
df = df.apply(lambda col: col.map(lambda x: x.replace('Â\xa0', ' ').replace('Â ', ' ').strip() if isinstance(x, str) else x))
```

```python
print(df['Engine'].unique())
```

```python
print("Duplikati:", df.duplicated().sum())

print("Jedinstveni automobili:", df[['Company', 'Model', 'Engine', 'Transmission', 'Color', 'Body Style']].drop_duplicates().shape[0])

print("Jedinstveni dealeri:", df[['Dealer_Name', 'Dealer_Region']].drop_duplicates().shape[0])

print("Jedinstveni kupaci:", df[['Customer Name', 'Gender', 'Annual Income']].drop_duplicates().shape[0])

print("Jedinstveni datumi:", df['Date'].nunique())
```

```python
dim_car = df[['Company', 'Model', 'Engine', 'Transmission', 'Color', 'Body Style']].drop_duplicates().reset_index(drop=True)
dim_car.insert(0, 'car_id', range(1, len(dim_car) + 1))
print(dim_car.shape)
dim_car.head()
```

```python
dim_dealer = df[['Dealer_Name', 'Dealer_Region']].drop_duplicates().reset_index(drop=True)
dim_dealer.insert(0, 'dealer_id', range(1, len(dim_dealer) + 1))
print(dim_dealer.shape)
dim_dealer.head()
```

```python
dim_customer = df[['Customer Name', 'Gender', 'Annual Income']].drop_duplicates().reset_index(drop=True)
dim_customer.insert(0, 'customer_id', range(1, len(dim_customer) + 1))
print(dim_customer.shape)
dim_customer.head()
```

```python
df['Date'] = pd.to_datetime(df['Date'])

dim_date = df[['Date']].drop_duplicates().reset_index(drop=True)
dim_date.insert(0, 'date_id', range(1, len(dim_date) + 1))
dim_date['year']  = dim_date['Date'].dt.year
dim_date['month'] = dim_date['Date'].dt.month
dim_date['day']   = dim_date['Date'].dt.day
print(dim_date.shape)
dim_date.head()
```

```python
F_car = df.merge(dim_car[['car_id','Company','Model','Engine','Transmission','Color','Body Style']], 
                on=['Company','Model','Engine','Transmission','Color','Body Style'])

F_car = F_car.merge(dim_dealer[['dealer_id','Dealer_Name','Dealer_Region']], 
                  on=['Dealer_Name','Dealer_Region'])

F_car = F_car.merge(dim_customer[['customer_id','Customer Name','Gender','Annual Income']], 
                  on=['Customer Name','Gender','Annual Income'])

F_car = F_car.merge(dim_date[['date_id','Date']], 
                  on='Date')

F_car_sale = F_car[['Sale_ID','car_id','dealer_id','customer_id','date_id','Price ($)']].copy()
F_car_sale.rename(columns={'Sale_ID': 'F_car_id', 'Price ($)': 'price'}, inplace=True)

print(F_car_sale.shape)
F_car_sale.head(15)
```

```python
F_car_sale.shape
```

```python
dim_car.to_csv('../data/dim_car.csv', index=False)
dim_dealer.to_csv('../data/dim_dealer.csv', index=False)
dim_customer.to_csv('../data/dim_customer.csv', index=False)
dim_date.to_csv('../data/dim_date.csv', index=False)
F_car_sale.to_csv('../data/fact_sale.csv', index=False)
```
